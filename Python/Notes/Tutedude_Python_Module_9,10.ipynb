```python
def create_identity_matrix_using_list_comprehension(n):
  return [[(1 if row == column else 0) for column in range(n)] for row in range(n)]
def print_matrix(matrix):
  for row in matrix:
    print(row)

I4 = create_identity_matrix_using_list_comprehension(4)
print_matrix(I4)
```

```python
def create_identity_matrix_using_nested_loops(n):
  matrix = []
  for row_index in range(n):
    row = []
    for column_index in range(n):
      entry = 1 if row_index == column_index else 0
      row.append(entry)
    matrix.append(row)

  return matrix

I5 = create_identity_matrix_using_nested_loops(5)
print_matrix(I5)
```

```python
def add_matrices(a,b):
  if len(a) != len(b) or len(a[0]) != len(b[0]):
    print(" The matrices should of the same size")
    return -1
  sum = []
  for i in range(len(a)):
    row = []
    for j in range(len(a[i])):
      row.append(a[i][j]+b[i][j])
    sum.append(row)
  return sum

print_matrix(add_matrices(I4,I4))
```

### Type of arguments to a function
- positional parameter in function definition: The parameters for which we do not specify a default value
- default or optional parameter. These are parameters for which a default value has been specified. if the arguments are not specified in the function call for this parameter then their default value is used.
- named or keyword arguments: this only makes sense for a function call. If we specify the name of the argument in the function call then it is a named argument. Also with named arguments the ordering does not matter.
- all positional parameters must precede the default parameters in the function definition.
- in the function call there are positional arguments and keyword arguments.
- a positional argument in the function call may be a positional parameter or a default parameter in the function definition.
- a keyword argument in function call may be a positional parameter or a default parameter in function definition.
- all positional arguments in the function call should be specified before keyword arguments.
- the ordering of the keyword arguments in the function call does not matter.
- **(Conclusion)**
  - in the function call all the positional arguments are specified before the keyword arguments.
  - after the positional arguments have been specified in the function call, the keyword arguments can be specified in any order and they may skip specifying some. But in the function call arguments for all the non default parameters must be specified. An argument for a non default parameter may be specified with a keyword.An argument for a default parameter may not be specified in the function call.

## Recursive functions
- Any recursive function has twomain parts:
  - (Terminal Condition) A condition when it stops calling itself.
  - (it calls itself) it expresses the value of the function in terms of the value of the function at a different argument

```python
# Factorial without recursion

def factnr(n):
  """
  n: positive integer
  return: n!

  calculates the factorial without using recursion
  """

  if n < 0 or not isinstance(n,int):
    return "Error: n should be a positive integer"
  factorial = 1
  while(n > 1):
    factorial *= n
    n -= 1
  return factorial
for n in range(11):
  print(f"{n}! = {factnr(n)}")
```

```python
def factr(n):
  """
  n: positive integer
  return: n!

  calculates the factorial using recursion
  """
  if n < 1 or not isinstance(n,int):
    return "Error: n should be a positive integer"
  return 1 if n == 1 else n * factr(n-1)

for n in range(11):
  print(f"{n}! = {factr(n)}")
```

```python
# generate the Fibonacci sequence

def F(n):
  """
  n: a nonegative integer
  return: the nth Fibonaacci number

  Generates the nth Fibonacci number using recursion
  """
  if not isinstance(n,int) or n < 0:
    return "Error: n must be a nonegative integer"
  if n==0 or n==1:
    return n
  else:
    return F(n-1) + F(n-2)
N = 30
for n in range(N):
  print(f"F({n}) = {F(n)}")
```

```python
# fibonacci numbers using lists
def F(n):
  """
  n: nonegative integer
  return: the nth Fibonacci number

  calculates the fibonacci numbers using an array and reusing any intermediate values
  """
  if not isinstance(n,int) or n < 0:
    return "Error: n must be a nonegative integer"

  if n < 2:
    return n

  f = [0]*(n+1)
  f[0], f[1] = 0, 1

  k = 1
  while k < n:
    k = k+1
    f[k] = f[k-1] + f[k-2]
  return f[k]
# N=10000 quite easy
N = 10
for n in range(N):
  print(f"F({n}) = {F(n)}")
```

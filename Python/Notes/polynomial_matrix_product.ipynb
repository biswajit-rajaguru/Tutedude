# A polynomial class, with polynomial addition and polynomial multiplication 

```python
'''
Polynomial class
p = Polynomial(coefficients = [0])
the constructor takes the list of coefficients p_0,p_1,ldots,p_d where d is the degree of the polynomial
'''

class Polynomial:

    @property
    def degree(self):
        return len(self.coefficients) - 1
        
    def __init__(self, coefficients = None):
        if coefficients is None:
            coefficients = [0]
        else:
            coefficients = list(coefficients)

        
        while len(coefficients)> 1 and coefficients[-1] == 0:
            coefficients.pop()
        
        self.coefficients = coefficients

    # p_1 + p_2 

    def __add__(self, b):
        if isinstance(b, (int, float)):
            b = Polynomial([b])
        if self.degree > b.degree:
            return Polynomial([self.coefficients[i] + b.coefficients[i] for i in range(len(b.coefficients))] + self.coefficients[len(b.coefficients):])
        else:
            return Polynomial([b.coefficients[i] + self.coefficients[i] for i in range(len(self.coefficients))] + b.coefficients[len(self.coefficients):])

    def __radd__(self, b):
        return self + b

    

    def __sub__(self, b):
        if isinstance(b, (int, float)):
            b = Polynomial([b])
        bminus = Polynomial([-1*c for c in b.coefficients])
        return self + bminus

    def __rsub__(self, b):
        if isinstance(b, (int, float)):
            b = Polynomial([b]) 
        return b - self 
        
    
    
    def __mul__(self,b):
        if isinstance(b, (int, float)):
            b = Polynomial([b])
        #print(f"Multiplying {self} {b} {'-'*40}")
        product = Polynomial()
        d = 0
        #print(f"product: {product}")
        for i in self.coefficients:
            #t = Polynomial([0]*d + [i*x for x in b.coefficients])
            #print(f"i: {i}, b: {b}, d: {d}, t: {t}")
            product = product + Polynomial([0]*d + [i*x for x in b.coefficients])
            d = d + 1
            #print(f"product: {product}")
        return product


    def __rmul__(self,b):
        return self * b
    
    
    
    
    def __str__(self):
        return f"poly{self.coefficients}"





    

if __name__ == "__main__":
    
    a = Polynomial([1,0])
    b = Polynomial([0,1])
    c = Polynomial([0])
    d = Polynomial([1])
    
    #e = Polynomial([0,1])
    #f = Polynomial([0,0,0,0,1])
    
    print(f"a:{a} - b:{b} = {a - b}")
    print(f"a:{a} - c:{c} = {a - c}")
    print(f"a:{a} - c:{d} = {a - d}")
    
    #print(f"a:{e} + c:{f} = {e*f}")
```

# A Matrix class with Matrix addition, subtraction, multiplication.

```python
# A matrix class with addition, subtraction, multiplication.

def generate_an_array_of_shape(shape, element = 0):
    return [[element]*shape[1] for i in range(shape[0])]

class Matrix:
    def __init__(self,shape = [1,1], entries = None):
        self.shape = shape
        if not entries:
            entries = generate_an_array_of_shape(shape,0)
        if len(entries) != shape[0] or len(entries[0]) != shape[1]:
            raise Exception("Error: Shape of {entries} is not {shape}")
        self.entries = entries
    
    
    def __add__(self,b):
        if self.shape != b.shape:
            raise Exception("Error(Matrix addition) {self}, {b} are not of the same shape")
        sum = Matrix(shape = self.shape)
        #print(f"sum: {sum}")
        for i in range(self.shape[0]):
            for j in range(self.shape[1]):
                sum.entries[i][j] = self.entries[i][j] + b.entries[i][j]
                #print(f"{sum.entries[i][j]} = {self.entries[i][j]} + {b.entries[i][j]}")
                #print(sum)
        return sum
    
    def __sub__(self, b):
        bminus = Matrix(b.shape)
        for i in range(b.shape[0]):
            for j in range(b.shape[1]):
                bminus.entries[i][j] = b.entries[i][j] * (-1)
        return self + bminus


    def __mul__(self, b):
        if self.shape[1] != b.shape[0]:
            raise Exception("Error(Matrix Multiplication): Row size of A is not equal to column size of B when evaluating A*B")
        product = Matrix([self.shape[0], b.shape[1]])
        for i in range(self.shape[0]):
            for j in range(b.shape[1]):
                s = 0
                for k in range(self.shape[1]):
                    s = s + (self.entries[i][k] * b.entries[k][j])
                product.entries[i][j] = s
        return product
    
    
    def __str__(self): 
        s = "\n-\n"
        for row in self.entries: 
            s += "["
            for entry in row:
                s += f"{entry}, "
            s += "]\n"
        return s

'''    
a = Matrix([3,3],[[1,1,1],[1,2,3],[0,1,0]]) 
b = Matrix([3,3],[[-1,0,-1],[1,2,3],[0,1,-1]]) 
print(f"{a}, {b}")
print(f"a + b = {a+b}")
print(f"a - b = {a-b}")
print(f"b - a = {b-a}")
print(f"a * b = {a*b}")
print(f"b * a = {b*a}")

x = Polynomial([0,1])
a = Matrix([2,2],[[x,x],[x,x]])
print(a+a*a + a*a*a)
'''
#print(generate_an_array_of_shape([5,2],"*"))
```

# The id function

## The id function.

- Each object in python that is strored in memory has a unique id. In CPython it used to be th memeory address, but it's exact meaning can be implementation dependent.
- Each immutatble data like a string, a number an integer has a a id attached to it. The variable points to it.
- For example
```python
l = 10
print(id(l))
l = l+1
print(id(l)) # most probably would be different.
```

- so the id is associated with the value. The id associated to a variable depends upon the value of the variable.
- I guess this is how dynamic typing works.
- same with strings, tuples etc
- `a is b` is same as `id(a) == id(b)` tests if a and b have the same id.
- lists are mutable. when lists are mutated for example by by appending a value the id **stays the same**.
- if two variables a, both contain the value 10, their ids may be the same as they may be pointing to the "same" value.
- Note the difference between "same" and "equal"

## Frozen Set

- Frozen set is a set like object which is immutable
- We can make a set a frozenset using frozenset(x), where x is a iterable
- when printed they are printed they are printed with a prefix frozenset.
- set expression containing only frozen sets returns a frozen set.
- a set operation containg normal sets and frozen sets returns normal sets.
- unlike sets a frozen set can be an element of a set.
- frozen sets do not have methods like add which mutate the set.


````python
a= frozenset("hello")
print(a)
x = {a}
y = x | a
print(x,y)
````

````python
# when printed they are printed they are printed with a prefix frozenset.
a = frozenset("hello")
print(a)
````

# Shallow and Deep copy

### Shallow Copy

````python


l1 = [1,2,[1,2,3],4]
print(l1)
l2 = l1
print(f"id of l1 is {id(l1)} while id of l2 is {id(l2)}")
````

````python
# using copy.copy

import copy

l2 = copy.copy(l1)
print(f"id of l1 is {id(l1)} while id of l2 is {id(l2)}")
print(f"id of l1 is {id(l1[2])} while id of l2 is {id(l2[2])}")
````

# Module-11-Files

### modes r w x a t b
- r = rt, rb only read File must exist or FileNotFoundError
- w = wt, wb only write. if file exists truncate it and no error if file does not exist
- x = xt, xb create a new file. if file exists we will have a `FileExistsError`. Check against accidently overwriting existing files
- a = zt, ab append to end end of file. File pointer at the end. Creates a file if it does not exist.
- t for text file and text mode
- b for binary file and binary mode
- r+ = r+t, r+b read and write both. File must exist or FileNotFoundError. File Contents are preserved. File pointer at begining.
- w+ = w+t, w+b write and read. If file exists truncate it first. If file does not exist then create it and no error.

### A good practice for editing/updating files
- For editing a file. Open the file in `r` mode. Copy contents into memory. Then close it. Edit the contents in memory. Then open the file in `w` mode. write the contents. then close it.
- the above workflow is safer and better than `r+` or `w+`.
- If the file should be new, then use `x` mode to protect against overwriting an existing file by mistake.

### read() on a file that is open for read access reads the entire contents of the file and returns it as string. 

```python
f1 = open("test-2.txt", "rt")
t1 = f1.read()
f1.close()

f2 = open("test-1.txt", "rt")
t2 = f2.read()
f2.close()

f3 = open("test-3.txt","xt")
t = t1 + t2
f3.write(t)
f3.close()
```

```python
def cpfile(file1, file2):
    f1 = open(file1, "rt")
    t = f1.read()
    f1.close()
    f2 = open(file2, "xt")
    f2.write(t)
    f2.close()
    print(f"Copied {file1} to {file2}")
```

### read()
- read() : read entire file
- read(n) : read the first n characters
- readline()
- readlines()

```python
cpfile("test-3.txt","test-4.txt")
```

```python
def head(file, n=10):
    f = open(file, "rt")
    t = ""
    for _ in range(n):
        t += f.readline()
    f.close()
    return t

        

```

```python
print(f"the head of \"test-3.txt\" {head("test-3.txt")}")
```

```python
print(f"the head of \"test-3.txt\" {head("test-3.txt",4)}")
```

### 
- readline(): a single line.
    - if at EOF then return empty string.
    - '\n' for blank line
- readlines(): all the lines as a list.
  

```python
def readlinesf(file):
    f = open(file, "rt")
    lines = f.readlines()
    f.close()
    return lines
```

```python
lines = readlinesf("test-1.txt")
for line in lines:
    print(line.strip())
print(type(lines))
```

## with
- `with open(file,mode) as f:`
- We start a context under the guardianship of with,
- do our work in the context.
- then leave the context
- **the guardian takes care of cleanup operations.**
- For ope file operation, the cleanup is handling any exception and properly closing the file.

```python
file, mode = "test-2.txt", "rt"
with open(file, mode) as f:
    while (t := f.readline()) != '':
        print(t.strip())

    
```

### check if a file exists
- using os.path.exists
- path is a relative path from the cwd if it does not start with "/"
- if it starts with "/", then it is a absolute path
- using pathlib.Path.exists().
- for the pathlib module the path has the same format as the os module
  

```python
import os
filename = "test-4.txt"
if os.path.exists(filename):
    print(f"file {filename} exists.")
else:
    print(f"file {filename} does not exist.")
```

```python
# an example of a basic try-except

a = int(input("Enter a number: "))
b = int(input("Enter denominator: "))

try:
    print(f"{a}/{b} = {a/b}")
except:
    print("the denominator cannot be 0")
```

```python
# another example of try-except

while True:
    try:
        a = int(input("Enter a number: "))
    except ValueError as error:
        print("input has to be an integer")
        print(error, type(error))
        continue
    except Exception as error:
        print("Some error")
    else:
        break
    finally:
        print("Try Number input")
while True:
    try:
        b = int(input("Enter the divisor: "))
    except Exception as error:
        print("input has to be an integer")
        print(error, type(error))
        continue
    else:
        break
    finally:
        print("Try: Divisor input")



try:
    print(f"{a}/{b} = {a/b}")
except ZeroDivisionError as error:
    print("the denominator cannot be 0")
    print(f"{error} and type of error is {type(error)}")
else:
    print("No errors")
finally:
        print("Try Division")
```

### Error object
- except ZeroDivisionError as error
- type of error is <class 'ZeroDivisionError'>
### Default except
- `except:` when you do not want to catch the error
- `except Exception as error:` Exception is the base class for most errors and will catch most errors

### Raise exception
- raise ValueError("Error msg") etc
- raise Exception("Error msg") is the generic exception.

```python
while True:
    try:
        salary = float(input("salary: "))
        if salary < 0:
            raise ValueError("negative")
    except ValueError as error:
        if str(error) == "negative":
            print("Salary cannot be negative")
        else:
            print(error)
        continue
    else:
        break
    finally:
        pass

print(f"Salary: {salary}")
    
```

# json module
- ## serialisation: dict -> json === `json.dump(dict,fh,indent=4)`
- ## deserialisation: json -> dict === `json.load(fh)`

```python
from random import randint, random
import json
students = {"studen"+str(n):{"name":"name"+str(n),"marks":randint(1,1000),"percent":f"{random()*100:.2f}"+"%"} for n in range(10)}
for student in students:
    print(students[student])
with open("students.json", "wt") as f:
    json.dump(students, f, indent = 4)
```

## Updating Json file

```python
import json
with open("students.json", "r") as f:
    data = json.load(f)
print(type(data))
d = {"dummy":{"dumm1":{"dummy2":1}}}
data.update(d)
for key in data:
    print(f"{key}: {data[key]}")
d = {"dummy":{"dumm1":{"dummy2":1}}}
data.update(d)
with open("students.json", "w") as f:
    data = json.dump(data, f, indent = 4)
```

## pickle module

```python
import pickle
import json
with open("students.json", "rt") as f:
    data = json.load(f)
with open ("students.pkl", "wb") as f:
    pickle.dump(data, f)
    data["studen0"]["name"] = "Biswajit"
    pickle.dump(data,f)

```

## use try except to load data from a pickle file

```python
with open ("students.pkl", "rb") as f:
    while True:
        try:
            data = pickle.load(f)
            for item in data:
                if "name" in data[item]:
                    print(data[item]["name"])
        except EOFError as error:
            print("pickle finished loading data")
            break
            
```

```python
"""
import re
re.search(regex, string) => match object if match found else returns None
"""
import re

message = "The present version of Python is 3.13.5"
pattern = "13"
match_obj = re.search(pattern, message)
### A common use case to determine if match found
if match_obj:
    print("Match Found")
else:
    print("Match Not Found")

print(match_obj)
```

```python
import re

message = "The present version of Python is 3.13.5. Other previous versions are 3.09, 3.10, 3.11, 3.12."

print(f"[0-9][0-9]: {re.search('[0-9][0-9]', message)}")
print(f"[0-9][0-9].[0-9]: {re.search('[0-9][0-9].[0-9]', message)}")
print(f"[.][0-9][.]: {re.search('[.][0-9][.]', message)}")
print(f"[0-9][0-9]: {re.search('[0-9][0-9]', 'House number is 1 2 3 123/A')}")
print(f"[0-9][0-9][0-9]: {re.search('[0-9][0-9][0-9]', 'House number is 1 2 3 123/A')}")
print(f"[0-9]: {re.search('[0-9]', 'House number is 1 2 3 123/A')}")
```

```python
import re 
s = "Python is a programming language. My current version 3.13.5"
s1 = """Python is a programming language.
My current version 3.13.5"""
s2 = "Python is a programming language."
s3 = "randon.number@gmail.com hyui@math.ku.edu"
pat = r"[a-z][a-z]"
s4 = "dfksdfsdlkjflkdsjababababababbbb"
s5 = "catfood dogfood"
s6 = "2025-11-24"
s7 = "2025|11|24"
s8 = "Paris in the the spring"
s9 = "<v>sdjas</g>jdhjshdk<a>fdssdfs</a>dfsfhsjkdhfskjhfskdjasdhjhjsdj<b>djsfhskj</b>"
print(re.search(r"[a-z][a-z]",s))
print(re.search(r"\.\d",s))
print(re.search(r"[a-z]*\D ",s))
print(re.search(r"\S\s\S",s1))
print(re.search(r"...\n",s1)) 
print(re.search(r"[a-z]{10}",s1))
print(re.search(r"[a-z]{7}.$",s2))
print(re.search(r"(com)",s))
print(re.search(r"(ab){4}", s4))
print(re.search(r"ab{4}", s4))
print(re.search(r"(cat|dog)food", s5))
print(re.search(r"cat|dogfood", s5))
match1 = re.search(r"(\d{4})(?:-|\|)(\d{2})(?:-|\|)(\d{2})", s6)
print(f"{match1.group(0)}\n{match1.group(1)}\n{match1.group(2)}\n{match1.group(3)}\n{match1.groups()}")
match1 = re.search(r"(\d{4})(?:-|\|)(\d{2})(?:-|\|)(\d{2})", s7)
print(f"{match1.group(0)}\n{match1.group(1)}\n{match1.group(2)}\n{match1.group(3)}\n{match1.groups()}")
print(re.search(r"(\w+)\s+\1",s8).group(1))
print(re.search(r"<(\w+)>(?:.*)</\1>",s9))
```

## An example of transforming text using regex . using backreferences

```python
import re
s = "My birthday is on 2025-11-24"
pat = r"(\d{4})-(\d{2})-(\d{2})"
repl = r"\3/\2/\1"
re.sub(pat, repl, s)
```

```python
s = "Hi Biswajit Rajaguru"
pat1 = r".*\s[A-Za-z]*"
pat2 = r".*?\s[A-Za-z]*"
print(re.search(pat1,s))
print(re.search(pat2,s))
```

```python
# recalling
import re
s = "Today is 2025-11-29"
pat = r"(\d{4})-(\d{2})-(\d{2})"
repl = r"\3 \2 \1"
re.sub(pat,repl,s)
```

```python
import re
s = "Today is 2025-11-29"
pat = r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
repl = r"\3 \2 \1"
match = re.search(pat,repl,s)
year, month, date = match.group("year"), match.group("month")
```

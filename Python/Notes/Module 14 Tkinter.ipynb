```python
#DEMO of dataobjects in tkinter

import tkinter as tk

# 1. Initialize the main window
root = tk.Tk()
root.title("Mirrored Entries")

# 2. Create the shared Tkinter Variable
# This object will hold the text content for BOTH Entry fields.
shared_text = tk.StringVar()

## --- First Entry (Primary Input) ---
label1 = tk.Label(root, text="Type Here:")
label1.pack(pady=(10, 0))

entry1 = tk.Entry(
    root, 
    width=50, 
    textvariable=shared_text  # Link to the shared variable
)
entry1.pack(padx=20, pady=5)

## --- Second Entry (Mirror) ---
label2 = tk.Label(root, text="Mirrored Text:")
label2.pack(pady=(10, 0))

entry2 = tk.Entry(
    root, 
    width=50, 
    textvariable=shared_text,  # Link to the EXACT SAME variable
    fg="blue"  # Just to show it's a different widget
)
entry2.pack(padx=20, pady=5)

root.mainloop()
```

```python
import tkinter as tk
import tkinter.font as tkfont

def quit_calculator():
    root.destroy()


root = tk.Tk()
root.geometry("700x500")
root.title("Calculator layout")

class CalculatorButton(tk.Button):
    buttonFont = tkfont.Font(family = "Helvetica", size = 14, weight = "bold")
    def __init__(self, master = None, **kwargs):

        kwargs.setdefault("font", CalculatorButton.buttonFont)
        super().__init__(master, **kwargs)
                
        if "activebackground" not in kwargs:
            self.configure(activebackground=self.cget("bg"))

        if "activeforeground" not in kwargs:
            self.configure(activeforeground=self.cget("fg"))

        self.configure(
            relief="flat",
            highlightthickness=0,
            #takefocus=False
        )

def onResize(event):
    if event.widget is not root:
        return
    cell_height = event.height / 5
    sizebutton = max(1, int(cell_height * 0.3))
    sizedisplay = max(1, int(cell_height * 0.5))
    CalculatorButton.buttonFont.configure(size = sizebutton)
    displayFont.configure(size = sizedisplay)
    

root.bind("<Configure>", onResize)

root.rowconfigure(0, weight = 1)
root.rowconfigure(1, weight = 3)
root.columnconfigure(0, weight = 1)

display_background_color = "#b3ffec"
display_foreground_color = "#001a00"
keypad_background_color = "#cccccc"
key_background_color = "#8c8c8c"
key_foreground_color = "#e6e6e6"
resetkey_background_color = "white"
resetkey_foreground_color = "#444444"

display = tk.Frame(root, bg = display_background_color)
keypad = tk.Frame(root, bg = keypad_background_color)
display.grid(row = 0, column = 0, sticky = "nsew", padx = 5, pady = 5)
keypad.grid(row = 1, column = 0, sticky = "nsew", padx = 5, pady = 5)

display.rowconfigure(0, weight = 1)
display.columnconfigure(0, weight = 1)
displayFont = tkfont.Font(family = "Helvetica", size = 24)
display_label = tk.Label(
    display,
    text = "0",
    fg = display_foreground_color,
    bg = display_background_color,
    anchor = "e",
    font = displayFont)
display_label.grid(row = 0, column = 0, sticky = "nsew")

keypad.rowconfigure(0,weight = 1)
keypad.columnconfigure(0, weight = 1)
keypad.columnconfigure(1, weight = 1)

digits = tk.Frame(keypad, bg = keypad_background_color)
operators = tk.Frame(keypad, bg = keypad_background_color)
digits.grid(row = 0, column = 0, sticky = "nsew", padx = 5, pady = 5)
operators.grid(row = 0, column = 1, sticky = "nsew", padx = 5, pady = 5)

for i in range(4):
    digits.rowconfigure(i, weight = 1)
    operators.rowconfigure(i, weight = 1)

for j in range(3):
    digits.columnconfigure(j, weight = 1)
    operators.columnconfigure(j, weight = 1)
digits_buttons = {}
for i,text in enumerate("789456123"):
    b = CalculatorButton(digits, text = text, bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
    digits_buttons[text] = b
    row, column = i//3, i % 3
    
    b.grid(row = row, column = column, sticky = "nsew", padx = 5, pady = 5)

digits_buttons["0"] = CalculatorButton(digits, text = "0", bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
digits_buttons["0"].grid(row = 3, column = 0, columnspan = 2, sticky = "nsew", padx = 5, pady = 5)

digits_buttons["."] = CalculatorButton(digits, text = ".", bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
digits_buttons["."].grid(row = 3, column = 2, sticky = "nsew", padx = 5, pady = 5)

operators_buttons = {}
for i,text in enumerate(["M+", "M-","MR", "\u00d7","\u00f7", "MC"]):
    operators_buttons[text] = CalculatorButton(operators, text = text, bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
    row, column = i//3, i % 3
    operators_buttons[text].grid(row = row, column = column, sticky = "nsew", padx = 5, pady = 5)

operators_buttons["+"] = CalculatorButton(operators, text = "+", bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
operators_buttons["+"].grid(row = 2, column = 0, rowspan = 2, sticky = "nsew", padx = 5, pady = 5)

operators_buttons["-"] = CalculatorButton(operators, text = "-", bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
operators_buttons["-"].grid(row = 2, column = 1, sticky = "nsew", padx = 5, pady = 5)

operators_buttons["="] = CalculatorButton(operators, text = "=", bg = key_background_color, fg = key_foreground_color, command = quit_calculator)
operators_buttons["="].grid(row = 3, column = 1, sticky = "nsew", padx = 5, pady = 5)

operators_buttons["C"] = CalculatorButton(operators, text = "C", bg = resetkey_background_color, fg = resetkey_foreground_color, command = quit_calculator)
operators_buttons["C"].grid(row = 2, column = 2, sticky = "nsew", padx = 5, pady = 5)

operators_buttons["AC"] = CalculatorButton(operators, text = "AC", bg = resetkey_background_color, fg = resetkey_foreground_color, command = quit_calculator)
operators_buttons["AC"].grid(row = 3, column = 2, sticky = "nsew", padx = 5, pady = 5)



root.mainloop()
```

```python
import tkinter as tk
import tkinter.font as tkfont

root = tk.Tk()
root.geometry("400x300")

# shared font object
button_font = tkfont.Font(family="Helvetica", size=14, weight="bold")

# layout
for r in range(4):
    root.rowconfigure(r, weight=1)
for c in range(4):
    root.columnconfigure(c, weight=1)

buttons = []

for i in range(16):
    b = tk.Button(root, text=str(i), font=button_font)
    b.grid(row=i // 4, column=i % 4, sticky="nsew", padx=4, pady=4)
    buttons.append(b)

# resize handler
def on_resize(event):
    # estimate cell height
    rows = 4
    cell_height = event.height / rows

    # heuristic: font size proportional to cell height
    new_size = max(2, int(cell_height * 1))
    button_font.configure(size=new_size)

root.bind("<Configure>", on_resize)

root.mainloop()
```

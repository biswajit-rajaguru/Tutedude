# The reason it only output 4 instead of 1 2 3 4 is the following
- When we evaluate literals the out put is the literal itself, which is a value.
- the output of the above cell is actually the last return value returned by the interpreter. which is 4
- This is differentfrom using print statements to print things which are outputted into the stdout which is what we see.
- The pepl additionally also outputs the return value of the code in the code cell, so it seems as if 1 is outputting 1. Actually this is the behaviour of repl which is to putput the final return value.

```python
"""
Area of a triangle using Heron's formula
A = sqrt{s(s-a)(s-b)(s-c)}
"""

a = float(input("a: "))
b = float(input("b: "))
c = float(input("c: "))
s = (a+b+c)/2
A=pow(s*(s-a)*(s-b)*(s-c),1/2)
print("Area: ", A)
```

```python
"""
SI = P * (R/100) * T
"""

P = float(input("Principal: "))
T = float(input("Time(in decimals): "))
R = float(input("Annual Rate of Interest"))

SI = P * (R/100) * T
CI = P * (pow(1+(R/100), T) - 1)

print("Simple interest is: ", SI)
print("Compound interest is: ", CI)
```

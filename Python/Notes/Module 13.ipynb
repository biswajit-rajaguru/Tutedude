# Object Oriented programming in Python

- Objects are containers. they contain various data about an a particular "object". These data is stored in the member variables. There are called members or attributes of the objects. 
- they also contain methods which are procedures by which we interact with the object. These are called methods.
- For example gives a attribute. There are methods to get the value of the attribrute, as well a method to set the value of the attribrute.
- A class definition is where we define the structure of the class. Then we create object using the class definition or blue print. This is called instanciang or creating an instance of a class.
- Sometimes there are methods which are defined in the class which are independent of the object, instead they either act on all oobjects of a class or do not need a object to work.
- They are called class methods. in contrast to objects methods, which are exclusively concerned about the the particular object instance.
- attribrute:behavior :: member:method
- instance variables are variables local to the instance. each instance will have its own copy of them, even though they are defined in the class definition.
- class variable are those of which there is only copy for the class, all the instances share this variable, if they need to.

### The class definition defines a default object, and that is persistent  till maybe it is redefined. It has its own scope. and the variable defined within its scope are called class variables and objects are like children from that scope, so have access to the parents variables. 

- `__dict__` only shows instance variables
- class variables and class methods can be invoked using the class name as well as objects name

````python
class Student: 
    """
    This is a class student, to manage student records.
    """
    # class variables
    classname = "Student" 
    class_author = "Biswajit Rajaguru"

    @classmethod
    def classinfo(cls):
        print(f"This is the Student class method classinfo(). Its classname: {cls.classname} class_author: {cls.class_author}")

    @staticmethod
    def sm(name):
        print(f"Hello {name}. I am a static method")

    
    def __init__(self, name = "student", roll = "0000"):
        self.name = name
        self.roll = roll
        print("Creating Student object")
        
    def infoo(self):
        print(f"Instance variables: \nname: {self.name}, roll: {self.roll}\n")
        print(f"Class variables classname: {self.classname}, class_author: {self.class_author}")
    
s1 = Student("Munu", "0011")
print(s1.classname, s1.class_author)
print(s1.classinfo())
s1.infoo()

print("Using Class name", Student.class_author)
Student.classinfo()

s1.sm("Rokhima")
    
````

### `@classmethod` `@staticmethod` decorator 

### static methods
- neither bound to class or instance.
- they are put in the class definition for packaging purpose
- they donot get a internall argument like class or the object. So no cls or self as 1st arg.

````python
class Contact: 
    phoneDirectory = list()

    @classmethod
    def showc(cls):
        if not len(cls.phoneDirectory): 
            print("Contacts list is empty")
            return
        for entry in cls.phoneDirectory:
            entry.showo()
            

    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number
        Contact.phoneDirectory.append(self)
        
    def showo(self):
        print(f"Name: {self.name} Phone number: {self.phone_number}")

    @classmethod
    def searchc(cls, name="", phone_number=""):
        for contact in cls.phoneDirectory: 
            if contact.name == name or contact.phone_number == phone_number:
                print("match found")
                contact.showo()
                return
        print("Contact not found")
    
                

#for x in "strange":
#    Contact(x,x+"***")

while True:
    name = input("Name: ")
    if name == "quit":
        break
    phone_number = input("Phone Number: ")
    Contact(name, phone_number)


Contact.showc() 
print("<>|"*15)

Contact.searchc(phone_number="002")
Contact.searchc(name="Munu")
    
````

### base/parent class -> derived/child class

````python
class Vehicle:
    companyName = "Escorts"

    def __init__(self, nwheels, nseats, mileage): 
        self.nwheels = nwheels
        self.nseats = nseats
        self.mileage = mileage

    def get_details(self):
        return [self.nwheels, self.nseats, self.mileage]

v1 = Vehicle(4,4,10)
print(v1.get_details())

class Car(Vehicle):
    #def __init__(self, car_type = "sedan", wheel_size = "16"): 
    def __init__(self, car_type = "sedan", wheel_size = "16", nwheels = "4", nseats = "4", mileage = "15"): 
        self.car_type = car_type
        self.wheel_size = wheel_size
        #Vehicle.__init__(self, nwheels, nseats, mileage)
        super().__init__(nwheels, nseats, mileage)

c1 = Car()
print(c1.__dict__)
print(c1.get_details())
#help(Car)
````

### while initializing a child class the child class `__init__` should call `__init__` of the parent class as in 
- `Vehicle.__init__(self,v1,v2,v3)` or `super().__init__(4, 4, 15)`
- when using super() you do not pass self explicitly. 

````python

# example to demonstrate the use of 
# super().__init__(**kwargs) 


class D:
    def __init__(self, d, **kwargs):
        self.d = d
        print("Class D init called")
        super().__init__(**kwargs)

class A(D):
    X = "alpha"
    def __init__(self, a, **kwargs):
        self.a = a
        print("Class A init called")
        super().__init__(**kwargs)
        

class B:
    Y = "beta"
    def __init__(self, b, **kwargs):
        self.b = b
        print("Class B init called")
        super().__init__(**kwargs)

class C(A,B):
    def __init__(self, a = 1, b = 2, c = 3, d = 4):
        self.c = c
        print("Class C init called")
        super().__init__(a = a, b = b, d = d)
    
c = C()

print(C.X, C.Y, c.__dict__)
#help(C)
````

### `a+b = a.__add__(b)`

````python
# example of operator overloading

class vec2d: 
    def __init__(self, x = 0, y = 0): 
        self.x = x
        self.y = y
    def __repr__(self):
        return f"vec2d({self.x},{self.y})"
    def __add__(self, other):
        if isinstance(other, vec2d):
            return vec2d(x = self.x + other.x, y = self.y + other.y)
        elif isinstance(other, (int, float)):
            return vec2d(self.x + other, self.y + other)
        else:
            return NotImplemented
    def __radd__(self, other):
        return self.__add__(other)

v1 = vec2d(1,2)
v2 = vec2d(5,2.3)
print(f"{v1} + {v2} = {v1 + v2}")
print(f"{.9} + {v2} = {.9 + v2}")

    
````

### method overloading as in defining the same method with two different signatures or argument lists creates two versions of the method and the appropriate version is called that is appropriate for the arguments passed is NOT SUPPORTED in python. if  a method is redefined, then only the latest version executed by the script is active.  

````python
# example of method overriding

class A:
    def greet(self):
        return "I am in class A"
class B(A):
    pass
a = A()
b = B()
print(a.greet())
print(b.greet())
class C(A):
    def greet(self):
        return f"Super's greeting: {super().greet()}, but my greet is: I am in class C"
c = C()
print(c.greet())
````

### Abstract class
- it is a skeleton class, which defines the generic structure of objects in some family
- But it does not define implementations.
- For example we can have a generic shape class, and
- it will have generic attribrutes like name,
- and generic methods like find_area, display_shape, scale shape etc.
- Then we can derive a rectangle class from this, and implement the various methods and attribrutes for a rectangle
- A good practice is to use the ABC to create abstract classes because they force the concete classes to implement the necessary methods, before they are used.
    

````python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area():
        pass
class Rectangle(Shape):
    def __init__(self, length=10, width = 10):
        self.length = length
        self.width = width
    def area(self):
        return self.length*self.width
    
a = Rectangle(5,2)
print(a.area())
````

```python
def create_marks_dictionary():
    marks_dictionary = {}
    while True:
        print()
        text1 = "Creating marks list"
        print(f"{text1}\n{'='*len(text1)}\n")
        print("Enter student name and marks to add to the list or 'done' when there is no further input\n")
        name = input("Name: ")
        if name == 'done':
            print()
            break
            
        try:
            marks = float(input("Marks: "))
        except ValueError:
            print("Marks should be an integer")
            continue
        else:
            marks_dictionary[name] = marks
        
    return marks_dictionary

marks_dictionary = create_marks_dictionary()

import pickle

pickle_file_name = "marks_list.pkl"

with open(pickle_file_name, "wb") as f:
    pickle.dump(marks_dictionary, f)

print(f"saved marks dictionary in the file {pickle_file_name}")


with open(pickle_file_name, "rb") as f:
    try:
        marks_dictionary_1 = pickle.load(f)
    except EOFError as error:
        print(f"{pickle_file_name} does not contain the marks dictionary")
    else:
        print(f"read marks dictionary from {pickle_file_name}")
        while True:
            name = input("Enter the student's name: ")
            if name in marks_dictionary:
                print(f"{name}'s marks: {marks_dictionary_1[name]}")
            elif name == "done":
                print("exiting this system")
                break
            else:
                print("Student not found")

#for key in marks_dictionary:
#    print(f"{key}: {marks_dictionary[key]}")
```
